```python
import time
from pprint import pprint

from beetea.bt import Action, BehaviorTree, BTResponse, Condition, Sequence


battery = {"level": 80}


def deliver_package():
    print("Driving to the destination...")
    time.sleep(0.5)
    print("Package delivered!")
    return True


tree = BehaviorTree(
    Sequence((
        Condition(lambda: battery["level"] > 20, label="batteryLevel"),
        Action(deliver_package, label="deliverPackage"),
    ))
)

while tree.tick() == BTResponse.RUNNING:
    time.sleep(0.05)

print("Tree finished with:", tree.get_status().name)
print()
print(tree.get_program())
print()
pprint(tree.to_dict())
tree.halt()
```

```python
from beetea import parse_bt

program = """
behaviortree
  sequence
    #batteryLevel
    #deliverPackage
"""
parse_bt(program, symbols={
    "batteryLevel": lambda: Condition(lambda: battery["level"] > 20, label="batteryLevel"),
    "deliverPackage": lambda: Action(deliver_package, label="deliverPackage"),
})
```
